## Unclipped Bayesian Score Check

This sweep tests ranking with raw, unclipped Bayesian predictions. The goal is to check whether preserving out-of-range score ordering improves recommendations compared with clipping predictions back to the 1-10 MAL rating range before applying the uncertainty penalty.

```python
score_variant = "unclipped"
clip_predictions = False
weights_uncertainty = [12]
n_runs = 200
tuning_top_ks = [5, 10]
random_state = 42

eval_users = [
    "Cresherhsm", "Dadalt", "Emily3k", "Sturmx", "TheFunkyFetus",
    "YungJollof", "heemini", "kaninhoppning", "leoabx",
    "DanteNani2004", "KuiKuix", "LILITH_OG", "Lemmegoxd",
    "LuffyThiago", "Qhair", "SteelPax", "chekkit", "elienai_s",
    "ikodrmz", "Czerik", "DoomSlayer_OG", "Goril", "Irseus",
    "LocNguyen", "Opelo_Stradyon", "Stusbu", "azriqluqman",
    "chenamaty", "okami20",
]

from anime_evaluation import RankingMetricEvaluator

missing_users = [user for user in eval_users if user not in user_data]
if missing_users:
    raise KeyError(f"These eval users are missing from user_data: {missing_users}")

score_client = MALClient(client_id)
unclipped_user_scores = {
    user: score_client.get_scores(user_data[user])
    for user in eval_users
}

unclipped_per_user_metrics = []

for i, (username, scores) in enumerate(unclipped_user_scores.items(), start=1):
    print(f"[{i}/{len(unclipped_user_scores)}] username: {username}")
    try:
        ranking_evaluator = RankingMetricEvaluator(
            anime_df_scaled=anime_df_scaled,
            anime_df=anime_df,
            scores=scores,
            anime_data_client=anime_data_client,
            anime_data=anime_data,
            builder=builder,
            recommender=recommender,
        )
        _, ranking_summary = ranking_evaluator.tune_bayesian_uncertainty_ranking(
            weights=weights_uncertainty,
            n_runs=n_runs,
            top_ks=tuning_top_ks,
            random_state=random_state,
            clip_predictions=clip_predictions,
        )
    except ValueError as error:
        print(f"Skipping {username}: {error}")
        continue

    user_metrics = ranking_summary.rename(
        columns={"uncertainty_weight": "bayesian_uncertainty_weight"}
    )
    user_metrics.insert(0, "score_variant", score_variant)
    user_metrics.insert(1, "clip_predictions", clip_predictions)
    user_metrics["username"] = username
    user_metrics["score_count"] = len(scores)
    user_metrics["n_runs"] = n_runs
    user_metrics["random_state"] = random_state
    user_metrics["svd_explained_variance"] = builder.svd_explained_variance
    unclipped_per_user_metrics.append(user_metrics)

if not unclipped_per_user_metrics:
    raise ValueError("No users had enough usable ratings for unclipped scoring.")

unclipped_per_user_metrics_df = pd.concat(
    unclipped_per_user_metrics,
    ignore_index=True,
)

summary_group_cols = [
    "score_variant",
    "clip_predictions",
    "bayesian_uncertainty_weight",
    "k",
]
unclipped_average_metrics = (
    unclipped_per_user_metrics_df
    .groupby(summary_group_cols, as_index=False)
    .mean(numeric_only=True)
)
unclipped_average_metrics["users_evaluated"] = (
    unclipped_per_user_metrics_df["username"].nunique()
)
unclipped_average_metrics["n_runs"] = n_runs
unclipped_average_metrics["random_state"] = random_state
unclipped_average_metrics["svd_explained_variance"] = builder.svd_explained_variance
unclipped_average_metrics = unclipped_average_metrics.sort_values(
    ["k", "avg_precision_at_k", "avg_ndcg_at_k"],
    ascending=[True, False, False],
)

METRICS_DIR = PROJECT_ROOT / "metrics" / "multi_user_metrics"
METRICS_DIR.mkdir(parents=True, exist_ok=True)
summary_path = METRICS_DIR / "multi_user_unclipped_bayesian_score_check_200run_2026.csv"
per_user_path = METRICS_DIR / "multi_user_unclipped_bayesian_score_check_per_user_200run_2026.csv"

unclipped_average_metrics.to_csv(summary_path, index=False)
unclipped_per_user_metrics_df.to_csv(per_user_path, index=False)

print(f"Saved {summary_path.relative_to(PROJECT_ROOT)}")
print(f"Saved {per_user_path.relative_to(PROJECT_ROOT)}")
display(unclipped_average_metrics)
```
