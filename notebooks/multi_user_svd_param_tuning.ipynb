Testing SVD param for 30 users. 

```python
from anime_evaluation import RankingMetricEvaluator
from anime_features import AnimeFeatureBuilder

svd_component_results = []
svd_per_user_results = []

n_runs = 100
clip_predictions = False
components = [50, 100, 200, 300]
max_tfidf_features = 3000
weights_uncertainty = [14]
tuning_top_ks = [5, 10]
users_per_bin = 10
sample_random_state = 42

svd_user_bins = {
    "completed_20_100": [
        user for user, completed in user_len.items()
        if 20 < completed < 100 and user in user_data
    ],
    "completed_100_400": [
        user for user, completed in user_len.items()
        if 100 < completed < 400 and user in user_data
    ],
    "completed_400_plus": [
        user for user, completed in user_len.items()
        if completed > 400 and user in user_data
    ],
}

rng = np.random.default_rng(sample_random_state)
svd_sampled_users_by_bin = {}
for bin_name, bin_users in svd_user_bins.items():
    if len(bin_users) < users_per_bin:
        raise ValueError(
            f"Need {users_per_bin} users in {bin_name}, found {len(bin_users)}."
        )
    svd_sampled_users_by_bin[bin_name] = sorted(
        rng.choice(sorted(bin_users), size=users_per_bin, replace=False).tolist()
    )

svd_sampled_users = [
    user
    for bin_users in svd_sampled_users_by_bin.values()
    for user in bin_users
]

score_client = MALClient(client_id)
svd_user_scores = {
    user: score_client.get_scores(user_data[user])
    for user in svd_sampled_users
}

base_builder = AnimeFeatureBuilder(
    anime_data,
    max_tfidf_features=max_tfidf_features,
)
base_anime_df_num = base_builder.build_num_features()
base_anime_genres_df = base_builder.build_genre_features()
base_synopsis_tfidf, base_synopsis_anime_ids = base_builder.build_synopsis_features(
    fit_tfidf=True,
)

for component in components:
    print(component)
    builder = AnimeFeatureBuilder(
        anime_data,
        max_tfidf_features=max_tfidf_features,
        n_svd_components=component,
    )
    builder.tfidf = base_builder.tfidf

    synopsis_svd_df = builder.apply_svd(
        base_synopsis_tfidf,
        base_synopsis_anime_ids,
        fit_svd=True,
    )
    component_anime_df = pd.concat([
        base_anime_df_num.set_index("id"),
        base_anime_genres_df.set_index("anime_id"),
        synopsis_svd_df,
    ], axis=1).dropna()

    recommender = SimilarityRecommender()
    recommender.create_anime_vectors(component_anime_df)
    component_anime_df_scaled = recommender.anime_df_scaled

    component_user_metrics = []

    for username, scores in svd_user_scores.items():
        try:
            ranking_evaluator = RankingMetricEvaluator(
                anime_df_scaled=component_anime_df_scaled,
                anime_df=component_anime_df,
                scores=scores,
                anime_data_client=anime_data_client,
                anime_data=anime_data,
                builder=builder,
                recommender=recommender,
            )
            _, ranking_summary = ranking_evaluator.tune_bayesian_uncertainty_ranking(
                weights=weights_uncertainty,
                n_runs=n_runs,
                top_ks=tuning_top_ks,
                random_state=42,
                clip_predictions=clip_predictions,
            )
        except ValueError as error:
            print(f"Skipping {username} for component {component}: {error}")
            continue

        user_metrics = ranking_summary.rename(
            columns={"uncertainty_weight": "bayesian_uncertainty_weight"}
        )
        user_metrics["username"] = username
        user_metrics["completed_bin"] = next(
            bin_name
            for bin_name, bin_users in svd_sampled_users_by_bin.items()
            if username in bin_users
        )
        user_metrics["score_count"] = len(scores)
        user_metrics["component"] = component
        user_metrics["max_tfidf_features"] = max_tfidf_features
        user_metrics["clip_predictions"] = clip_predictions
        user_metrics["n_runs"] = n_runs
        user_metrics["svd_explained_variance"] = builder.svd_explained_variance

        component_user_metrics.append(user_metrics)
        svd_per_user_results.append(user_metrics)

    if not component_user_metrics:
        continue

    component_user_metrics_df = pd.concat(component_user_metrics, ignore_index=True)
    average_metrics = (
        component_user_metrics_df
        .groupby(["component", "bayesian_uncertainty_weight", "k"], as_index=False)
        .mean(numeric_only=True)
    )
    average_metrics["users_evaluated"] = component_user_metrics_df["username"].nunique()
    average_metrics["max_tfidf_features"] = max_tfidf_features
    average_metrics["clip_predictions"] = clip_predictions
    average_metrics["n_runs"] = n_runs
    average_metrics["svd_explained_variance"] = builder.svd_explained_variance

    svd_component_results.append(average_metrics)

    svd_component_summary = (
        pd.concat(svd_component_results, ignore_index=True)
        .sort_values(
            ["k", "avg_precision_at_k", "avg_ndcg_at_k"],
            ascending=[True, False, False],
        )
    )

svd_per_user_results_df = (
    pd.concat(svd_per_user_results, ignore_index=True)
    if svd_per_user_results
    else pd.DataFrame()
)

svd_sampled_users_by_bin, svd_component_summary
```
