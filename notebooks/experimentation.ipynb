Testing for every user

```python
from anime_evaluation import RankingMetricEvaluator
import random

# user_scores_trun = dict(random.sample(list(user_scores.items()), 150))

n_runs = 100
compare_top_ks = [5]
uncertainty_weight = 20
relevance_threshold = 6
user_client = MALClient(client_id)

per_user_results = []
skipped_users = []

for counter, user in enumerate(sorted_user_info, start=1):
    print(f"[{counter}/{len(sorted_user_info)}] {user}, {sorted_user_info[user]}")

    try:
        user_data = user_client.get_user_data(user)
        user_scores_loop = user_client.get_scores(user_data)

        score_values = np.array(list(user_scores_loop.values()), dtype=float)
        threshold_attempts = [
            ("standard", float(np.median(score_values) + 0.5 * np.std(score_values))),
            ("median", float(np.median(score_values))),
            ("mean", float(np.mean(score_values))),
        ]

        last_error = None
        for threshold_strategy, like_threshold in threshold_attempts:
            try:
                evaluator = RankingMetricEvaluator(
                    anime_df_scaled=anime_df_scaled,
                    anime_df=anime_df,
                    scores=user_scores_loop,
                    anime_data_client=anime_data_client,
                    anime_data=anime_data,
                    builder=builder,
                    recommender=recommender,
                    like_threshold=like_threshold,
                    relevance_threshold=relevance_threshold,
                )

                compare_results, compare_summary = evaluator.compare_models(
                    uncertainty_weight=uncertainty_weight,
                    n_runs=n_runs,
                    top_ks=compare_top_ks,
                    include_ridge_cv=False,
                    include_lasso=False,
                    include_elastic_net=False,
                    include_knn=True,
                    include_hybrid_model=False,
                    include_gradient_boosting=False,
                    knn_neighbors=20,
                    knn_weights="distance",
                    knn_metric="cosine",
                    random_state=42,
                    clip_predictions=True,
                )
                break
            except ValueError as error:
                last_error = error
                print(
                    f"{threshold_strategy} threshold failed "
                    f"({like_threshold:.3f}): {error}"
                )
        else:
            raise last_error

        model_labels = {
            "bayesian_ridge": "bayesian",
            "knn": "knn_baseline",
            "global_mean": "global_baseline",
        }
        metric_labels = {
            "avg_precision_at_k": "precision_at_k",
            "avg_ndcg_at_k": "ndcg_at_k",
        }
        average_metrics = compare_summary[
            compare_summary["model"].isin(model_labels)
        ].pivot_table(
            index="k",
            columns="model",
            values=list(metric_labels),
        ).reset_index()

        average_metrics.columns = [
            "k"
            if metric == "k"
            else f"{model_labels[model]}_{metric_labels[metric]}"
            for metric, model in average_metrics.columns
        ]

        average_metrics["username"] = user
        average_metrics["user_len"] = sorted_user_info[user]
        average_metrics["score_count"] = len(user_scores_loop)
        average_metrics["liked_count"] = len(evaluator.liked_eval)
        average_metrics["like_threshold"] = evaluator.like_threshold
        average_metrics["relevance_threshold"] = evaluator.relevance_threshold
        average_metrics["threshold_strategy"] = threshold_strategy
        average_metrics["score_mean"] = np.mean(score_values)
        average_metrics["score_std"] = np.std(score_values)
        average_metrics["score_median"] = np.median(score_values)
        average_metrics["liked_ratio"] = len(evaluator.liked_eval) / len(evaluator.rated_eval)

        _, _, _, debug_test_eval = evaluator._sample_ranking_split(random_state=42)
        average_metrics["hidden"] = len(debug_test_eval)
        average_metrics["hidden_relevant"] = int((debug_test_eval["relevance"] > 0).sum())
        average_metrics["hidden_strong_relevant"] = int((debug_test_eval["relevance"] == 2).sum())
        print(
            f"Like threshold: {evaluator.like_threshold}, "
            f"Hidden: {len(debug_test_eval)}, "
            f"Relevant hidden: {int((debug_test_eval['relevance'] > 0).sum())}"
        )

        print(average_metrics[['bayesian_precision_at_k',
                               'bayesian_ndcg_at_k',
                               'knn_baseline_precision_at_k',
                               'knn_baseline_ndcg_at_k',
                               'liked_count']])
        per_user_results.append(average_metrics.copy())

    except (ValueError, RuntimeError) as error:
        print(f"Skipping {user}: {error}")
        skipped_users.append({
            "username": user,
            "user_len": sorted_user_info[user],
            "reason": str(error),
        })

per_user_results_df = (
    pd.concat(per_user_results, ignore_index=True)
    if per_user_results
    else pd.DataFrame()
)
skipped_users_df = pd.DataFrame(skipped_users)
# _, _, _, debug_test_eval = evaluator._sample_ranking_split(random_state=42)
# print("rated_eval:", len(evaluator.rated_eval))
# print("like_threshold:", evaluator.like_threshold)
# print("liked_eval:", len(evaluator.liked_eval))
# print("hidden_per_split:", len(debug_test_eval))
# print(anime_df.shape)
# display(average_metrics)
```

Bayesion Ridge Regression

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, root_mean_squared_error
from sklearn.linear_model import BayesianRidge
import numpy as np

rated_items, anime_df, anime_vectors, anime_df_scaled, builder = anime_data_client.get_rated_items(
    user_scores=user_scores,
    anime_data=anime_data,
    builder=builder,
    recommender=recommender,
    anime_df=anime_df,
    anime_vectors=anime_vectors,
)

X = np.array([vec for _, vec, _ in rated_items])
y = np.array([score for _, _, score in rated_items], dtype=float)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=None
)

model = BayesianRidge()
model.fit(X_train, y_train)

pred, pred_std = model.predict(X_test, return_std=True)
pred = np.clip(pred, 1, 10)

mae = mean_absolute_error(y_test, pred)
rmse = root_mean_squared_error(y_test, pred)

baseline_pred = np.full_like(y_test, y_train.mean(), dtype=float)
baseline_mae = mean_absolute_error(y_test, baseline_pred)

print(f"MAE: {mae:.3f}")
print(f"RMSE: {rmse:.3f}")
print(f"Baseline MAE: {baseline_mae:.3f}")
print(f"Improvement vs baseline: {baseline_mae - mae:.3f}")
```

Hit Rate

```python
# Hide some high-rated anime, train on the rest, and see if the hidden likes show up near the top.
hit_rating_threshold = np.median(y) + 0.5 * np.std(y)
heldout_fraction = 0.25
top_ks = [5, 10, 20, 50, 100]

rated_eval = pd.DataFrame({
    "anime_id": [anime_id for anime_id, _, _ in rated_items],
    "score": [score for _, _, score in rated_items],
})

liked_eval = rated_eval[rated_eval["score"] >= hit_rating_threshold]
if len(liked_eval) < 2:
    raise ValueError("Need at least 2 high-rated anime to run a hit-rate holdout test.")

heldout_liked = liked_eval.sample(frac=heldout_fraction, random_state=None)
heldout_ids = set(heldout_liked["anime_id"])

train_eval = rated_eval[~rated_eval["anime_id"].isin(heldout_ids)]
train_ids = train_eval["anime_id"].tolist()

X_hit_train = anime_df_scaled.loc[train_ids].to_numpy()
y_hit_train = train_eval["score"].to_numpy(dtype=float)

hit_model = BayesianRidge()
hit_model.fit(X_hit_train, y_hit_train)

# Candidates are everything the model did not train on, including the hidden liked anime.
candidate_ids = [anime_id for anime_id in anime_df_scaled.index if anime_id not in set(train_ids)]
X_hit_candidates = anime_df_scaled.loc[candidate_ids].to_numpy()

hit_pred, hit_pred_std = hit_model.predict(X_hit_candidates, return_std=True)
hit_pred = np.clip(hit_pred, 1, 10)

title_by_id = {int(anime["id"]): anime["title"] for anime in anime_data.values()}
hit_recommendations = pd.DataFrame({
    "anime_id": candidate_ids,
    "title": [title_by_id.get(int(anime_id), "Unknown") for anime_id in candidate_ids],
    "actual_score": [user_scores.get(int(anime_id)) for anime_id in candidate_ids],
    "predicted_score": hit_pred,
    "uncertainty_raw": hit_pred_std,
})

uncertainty_clipped = hit_recommendations["uncertainty_raw"].clip(
    lower=hit_recommendations["uncertainty_raw"].quantile(0.05),
    upper=hit_recommendations["uncertainty_raw"].quantile(0.95),
)
if uncertainty_clipped.max() == uncertainty_clipped.min():
    hit_recommendations["uncertainty"] = 0.0
else:
    hit_recommendations["uncertainty"] = (
        (uncertainty_clipped - uncertainty_clipped.min())
        / (uncertainty_clipped.max() - uncertainty_clipped.min())
    )

hit_recommendations["ranking_score"] = (
    hit_recommendations["predicted_score"] - 4.5 * hit_recommendations["uncertainty"]
)
hit_recommendations["is_hidden_like"] = hit_recommendations["anime_id"].isin(heldout_ids)
hit_recommendations = hit_recommendations.sort_values("ranking_score", ascending=False)

hit_rows = []
for k in top_ks:
    top_k = hit_recommendations.head(k)
    hits = int(top_k["is_hidden_like"].sum())
    hit_rows.append({
        "k": k,
        "hits": hits,
        "heldout_likes": len(heldout_ids),
        "hit_rate": hits / len(heldout_ids),
        "precision_at_k": hits / k,
    })

hit_rate_results = pd.DataFrame(hit_rows)

print(f"Total rated anime: {len(rated_eval)}")
print(f"High-rated anime (score >= {hit_rating_threshold}): {len(liked_eval)}")
print(f"Hidden liked anime: {len(heldout_ids)}")
display(hit_rate_results)

# hit_recommendations[hit_recommendations["is_hidden_like"]].head(20)
```

Tuning Bayesian

```python
weights_uncertainty = np.array([0, 0.1, 0.25, 0.5, 0.75, 1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5, 5.5, 6, 7, 8])
n_runs = 50
tuning_top_ks = [5, 10]

tuning_rows = []
baseline_rows = []

for run in range(n_runs):
    heldout_liked = liked_eval.sample(frac=heldout_fraction, random_state=None)
    heldout_ids = set(heldout_liked["anime_id"])

    train_eval = rated_eval[~rated_eval["anime_id"].isin(heldout_ids)]
    train_ids = train_eval["anime_id"].tolist()

    X_tune_train = anime_df_scaled.loc[train_ids].to_numpy()
    y_tune_train = train_eval["score"].to_numpy(dtype=float)

    candidate_ids_tune = [
        anime_id
        for anime_id in anime_df_scaled.index
        if anime_id not in set(train_ids)
    ]
    X_tune_candidates = anime_df_scaled.loc[candidate_ids_tune].to_numpy()

    tune_model = BayesianRidge()
    tune_model.fit(X_tune_train, y_tune_train)

    tune_pred, tune_std = tune_model.predict(X_tune_candidates, return_std=True)
    tune_pred = np.clip(tune_pred, 1, 10)

    tune_recommendations = pd.DataFrame({
        "anime_id": candidate_ids_tune,
        "is_hidden_like": pd.Series(candidate_ids_tune).isin(heldout_ids).to_numpy(),
        "predicted_score": tune_pred,
        "uncertainty_raw": tune_std,
    })

    uncertainty_clipped = tune_recommendations["uncertainty_raw"].clip(
        lower=tune_recommendations["uncertainty_raw"].quantile(0.05),
        upper=tune_recommendations["uncertainty_raw"].quantile(0.95),
    )
    if uncertainty_clipped.max() == uncertainty_clipped.min():
        tune_recommendations["uncertainty"] = 0.0
    else:
        tune_recommendations["uncertainty"] = (
            (uncertainty_clipped - uncertainty_clipped.min())
            / (uncertainty_clipped.max() - uncertainty_clipped.min())
        )

    baseline_recommendations = tune_recommendations[["anime_id", "is_hidden_like"]].copy()
    baseline_recommendations["baseline_score"] = anime_df.loc[
        baseline_recommendations["anime_id"], "mean"
    ].to_numpy()
    baseline_recommendations = baseline_recommendations.sort_values(
        "baseline_score",
        ascending=False,
    )

    for k in tuning_top_ks:
        top_k = baseline_recommendations.head(k)
        hits = int(top_k["is_hidden_like"].sum())
        baseline_rows.append({
            "run": run + 1,
            "k": k,
            "hits": hits,
            "heldout_likes": len(heldout_ids),
            "hit_rate": hits / len(heldout_ids),
            "precision_at_k": hits / k,
        })

    for weight in weights_uncertainty:
        tune_recommendations["ranking_score"] = (
            tune_recommendations["predicted_score"]
            - weight * tune_recommendations["uncertainty"]
        )
        ranked = tune_recommendations.sort_values("ranking_score", ascending=False)

        for k in tuning_top_ks:
            top_k = ranked.head(k)
            hits = int(top_k["is_hidden_like"].sum())
            tuning_rows.append({
                "run": run + 1,
                "uncertainty_weight": weight,
                "k": k,
                "hits": hits,
                "heldout_likes": len(heldout_ids),
                "hit_rate": hits / len(heldout_ids),
                "precision_at_k": hits / k,
            })

bayesian_tuning_results = pd.DataFrame(tuning_rows)
baseline_tuning_results = pd.DataFrame(baseline_rows)

bayesian_tuning_summary = (
    bayesian_tuning_results
    .groupby(["uncertainty_weight", "k"])
    .agg(
        avg_precision_at_k=("precision_at_k", "mean"),
        std_precision_at_k=("precision_at_k", "std"),
        avg_hit_rate=("hit_rate", "mean"),
        std_hit_rate=("hit_rate", "std"),
        avg_hits=("hits", "mean"),
    )
    .reset_index()
    .sort_values(["k", "avg_precision_at_k"], ascending=[True, False])
)

display(bayesian_tuning_summary)

baseline_tuning_summary = (
    baseline_tuning_results
    .groupby("k")
    .agg(
        baseline_avg_precision_at_k=("precision_at_k", "mean"),
        baseline_std_precision_at_k=("precision_at_k", "std"),
        baseline_avg_hit_rate=("hit_rate", "mean"),
        baseline_std_hit_rate=("hit_rate", "std"),
        baseline_avg_hits=("hits", "mean"),
    )
    .reset_index()
)

display(baseline_tuning_summary)

best_uncertainty_weights = (
    bayesian_tuning_summary
    .sort_values(["k", "avg_precision_at_k", "avg_hit_rate"], ascending=[True, False, False])
    .groupby("k")
    .head(1)
)
best_uncertainty_weights = best_uncertainty_weights.merge(
    baseline_tuning_summary,
    on="k",
    how="left",
)
best_uncertainty_weights
```

## Numeric Feature Multicollinearity Check

This check explains why the numeric feature builder keeps `mean`, `popularity`, `num_scoring_users`, and `num_episodes`, while dropping the MAL status-count fields and `rank`. Most of the dropped fields are overlapping popularity/activity signals, so keeping all of them would double-count the same information.

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

numeric_candidates = pd.DataFrame(anime_data.values())

statistics_df = pd.json_normalize(numeric_candidates["statistics"])
statistics_df = statistics_df.rename(columns={
    "num_list_users": "statistics_num_list_users",
    "status.watching": "watching",
    "status.completed": "completed",
    "status.on_hold": "on_hold",
    "status.dropped": "dropped",
    "status.plan_to_watch": "plan_to_watch",
})
statistics_df = statistics_df.apply(pd.to_numeric, errors="coerce")

numeric_candidates = pd.concat(
    [numeric_candidates.drop(columns=["statistics"]), statistics_df],
    axis=1,
)

numeric_columns = [
    "mean",
    "rank",
    "popularity",
    "num_list_users",
    "num_scoring_users",
    "num_episodes",
    "statistics_num_list_users",
    "watching",
    "completed",
    "on_hold",
    "dropped",
    "plan_to_watch",
]

numeric_analysis_df = (
    numeric_candidates[numeric_columns]
    .apply(pd.to_numeric, errors="coerce")
    .dropna()
)

numeric_analysis_df.head()
```

```python
corr = numeric_analysis_df.corr(method="spearman")

high_corr_pairs = []
for i, left_col in enumerate(corr.columns):
    for right_col in corr.columns[i + 1:]:
        correlation = corr.loc[left_col, right_col]
        if abs(correlation) >= 0.85:
            high_corr_pairs.append({
                "feature_1": left_col,
                "feature_2": right_col,
                "spearman_corr": correlation,
            })

high_corr_pairs = pd.DataFrame(high_corr_pairs).sort_values(
    "spearman_corr",
    key=lambda col: col.abs(),
    ascending=False,
)

display(corr.round(3))
high_corr_pairs
```

```python
def calculate_vif(df):
    rows = []
    for target_col in df.columns:
        X = df.drop(columns=[target_col]).to_numpy(dtype=float)
        y = df[target_col].to_numpy(dtype=float)

        model = LinearRegression()
        model.fit(X, y)
        r_squared = model.score(X, y)

        vif = np.inf if np.isclose(1 - r_squared, 0) else 1 / (1 - r_squared)
        rows.append({
            "feature": target_col,
            "r_squared_from_other_features": r_squared,
            "vif": vif,
        })

    return pd.DataFrame(rows).sort_values("vif", ascending=False)

reduced_numeric_analysis_df = numeric_analysis_df.drop(
    columns=[
        "completed",
        "on_hold",
        "statistics_num_list_users",
        # "watching",
        # "dropped",
        # "plan_to_watch",
        "num_list_users",
        # "num_scoring_users",
        "rank"
    ],
    errors="ignore",
)

vif_results = calculate_vif(reduced_numeric_analysis_df)
vif_results
```
